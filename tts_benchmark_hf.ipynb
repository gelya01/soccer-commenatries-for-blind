```python
import json
import time
from pathlib import Path

import numpy as np
import pandas as pd
import soundfile as sf
import torch
from IPython.display import Audio, display
from qwen_tts import Qwen3TTSModel
```

```python
# Валидный Qwen3-TTS id
MODEL_ID = "Qwen/Qwen3-TTS-12Hz-1.7B-CustomVoice"

# Выбери мужской голос из supported_speakers после загрузки модели
SPEAKER = 'aiden'
LANGUAGE = 'Russian'
INSTRUCT = 'Deliver a live soccer call in Russian: fast, crisp, and breathless, with disbelief turning into controlled panic as the danger builds, emphasizing key moments with sharp, short phrases.'

SILENCE_SEC = 0.10
DEDUPLICATE = True
```

## Kaggle Run

Этот блок нужен, если ноутбук запущен в Kaggle через VS Code remote kernel. Он перенастраивает пути и позволяет прогнать TTS прямо в Kaggle `working`.

```python
import os
from pathlib import Path

IS_KAGGLE = Path('/kaggle').exists() or 'KAGGLE_KERNEL_RUN_TYPE' in os.environ
print('IS_KAGGLE =', IS_KAGGLE)

if IS_KAGGLE:
    KAGGLE_INPUT = Path('/kaggle/input')
    KAGGLE_WORKING = Path('/kaggle/working')
    print('Kaggle input dir:', KAGGLE_INPUT)
    print('Kaggle working dir:', KAGGLE_WORKING)
else:
    print('Не Kaggle среда, этот блок можно пропустить.')
```

```python
# Если commentary_log.txt загружен в Kaggle dataset, укажи путь здесь.
# Примеры:
# INPUT_PATH = Path('/kaggle/input/my-dataset/commentary_log.txt')
# INPUT_PATH = next(Path('/kaggle/input').rglob('commentary_log.txt'))

if IS_KAGGLE:
    found = list(Path('/kaggle/input').rglob('commentary_log.txt'))
    if found:
        INPUT_PATH = found[0]
        OUT_DIR = Path('/kaggle/working/tts_qwen_phrases')
        MERGED_WAV = Path('/kaggle/working/commentary_tts_qwen_male_commentator.wav')
        CSV_PATH = Path('/kaggle/working/tts_qwen_phrase_benchmark.csv')
        OUT_DIR.mkdir(parents=True, exist_ok=True)
        print('INPUT_PATH =', INPUT_PATH)
        print('OUT_DIR =', OUT_DIR)
        print('MERGED_WAV =', MERGED_WAV)
        print('CSV_PATH =', CSV_PATH)
    else:
        raise FileNotFoundError('Не найден commentary_log.txt внутри /kaggle/input')
else:
    print('Не Kaggle среда, локальные пути остаются без изменений.')
```

### Что запускать в Kaggle

Если ты уже подключена к Kaggle kernel через VS Code, просто выполни ноутбук сверху вниз. После этого результаты будут лежать в `/kaggle/working`.

```python
display(Audio(filename=MERGED_WAV.as_posix()))
```

```python
texts = []
with INPUT_PATH.open('r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        obj = json.loads(line)
        t = str(obj.get('commentary', '')).strip()
        if t:
            texts.append(t)

if DEDUPLICATE:
    seen = set()
    uniq = []
    for t in texts:
        if t not in seen:
            seen.add(t)
            uniq.append(t)
    texts = uniq

print(f'Фраз: {len(texts)}')
print('Первая фраза:', texts[0][:160] + ('...' if len(texts[0]) > 160 else ''))
```

```python
device_map = 'cuda:0' if torch.cuda.is_available() else 'cpu'
dtype = torch.bfloat16 if torch.cuda.is_available() else torch.float32
print('CUDA:', torch.cuda.is_available())

model = Qwen3TTSModel.from_pretrained(
    MODEL_ID,
    device_map=device_map,
    dtype=dtype,
    # attn_implementation="flash_attention_2",
)

supported_speakers = model.get_supported_speakers() if callable(getattr(model, 'get_supported_speakers', None)) else None
print('Model loaded:', MODEL_ID)
print('Speaker:', SPEAKER, '| Language:', LANGUAGE)
print('Instruct:', INSTRUCT)
if supported_speakers is not None:
    print('Supported speakers:', supported_speakers)
    print('Если голос не нравится, возьми мужской id из списка выше и замени SPEAKER.')
```

```python
rows = []
parts = []
sr_global = None
global_start = time.perf_counter()

for i, text in enumerate(texts, start=1):
    # t0 = time.perf_counter()
    wavs, sr = model.generate_custom_voice(
        text=text,
        language=LANGUAGE,
        speaker=SPEAKER,
        instruct=INSTRUCT,
    )
    # elapsed = time.perf_counter() - t0

    audio = wavs[0]
    if audio.dtype != np.float32:
        audio = audio.astype(np.float32)

    # peak = float(np.max(np.abs(audio))) if len(audio) else 0.0
    # if peak > 0:
    #     audio = audio / peak

    duration = len(audio) / sr if len(audio) else 0.0
    # rtf = elapsed / duration if duration > 0 else np.inf

    wav_path = OUT_DIR / f'phrase_{i:03d}.wav'
    sf.write(wav_path.as_posix(), audio, sr)

    rows.append({
        'idx': i,
        'chars': len(text),
        # 'time_sec': elapsed,
        'audio_sec': duration,
        # 'rtf': rtf,
        'wav_path': wav_path.as_posix(),
        'text': text,
    })

    parts.append(audio)
    if i < len(texts):
        parts.append(np.zeros(int(sr * SILENCE_SEC), dtype=np.float32))
    sr_global = sr

total_wall = time.perf_counter() - global_start

merged = np.concatenate(parts) if parts else np.array([], dtype=np.float32)
merged_sec = (len(merged) / sr_global) if (len(merged) and sr_global) else 0.0
if sr_global is not None:
    sf.write(MERGED_WAV.as_posix(), merged, sr_global)

df = pd.DataFrame(rows)
df.to_csv(CSV_PATH, index=False, encoding='utf-8')

print(f'Сохранено merged audio: {MERGED_WAV}')
print(f'Сохранено per-phrase csv: {CSV_PATH}')
print(f'Общее время синтеза: {total_wall:.3f} sec')
print(f'Общая длительность аудио: {merged_sec:.3f} sec')
print(f'RTF total: {(total_wall / merged_sec) if merged_sec > 0 else np.inf:.3f}')
```

```python
display(df[['idx', 'chars', 'time_sec', 'audio_sec', 'wav_path']])
print('mean time_sec:', df['time_sec'].mean())
print('p50 time_sec:', df['time_sec'].quantile(0.5))
print('p95 time_sec:', df['time_sec'].quantile(0.95))
print('max time_sec:', df['time_sec'].max())
```
